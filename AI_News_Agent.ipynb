```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

# Initialize the model
model = init_chat_model(model="gemini-3.5-flash-lite", model_provider="google_genai")
```

```python
from langchain_community.tools import DuckDuckGoSearchRun

# Set up the search tool
search_tool = DuckDuckGoSearchRun()
tools = [search_tool]
```

```python
from langchain.agents import initialize_agent, AgentType

# Create the agent using the model and the search tool
agent = initialize_agent(
    tools,
    model,
    agent=AgentType.CHAT_ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)
```

```python
# Run the agent to get the latest AI news
response = agent.run("What are the latest major announcements in generative AI this week?")
print(response)
```
